```python
import re

def clean_text_for_rnn(series: pd.Series) -> pd.Series:
    """
    Быстрая минимальная очистка 400k строк для RNN + wiki-words.
    """
    return (
        series.astype(str)
        # 1. Приведение к нижнему регистру
        .str.lower()
        # 2. Удаление HTML-тегов
        .str.replace(r'<[^>]+>', ' ', regex=True)
        # 3. Удаление URL-ссылок
        .str.replace(r'https?://\S+|www\.\S+', ' ', regex=True)
        # 4. Оставляем только буквы, одиночные кавычки и основные знаки (!)
        .str.replace(r"[^a-z'\!\?\s]", ' ', regex=True)
        # 5. Схлопывание множественных пробелов
        .str.replace(r'\s+', ' ', regex=True)
        .str.strip()
    )
```

```python
df['text'] = clean_text_for_rnn(df['text'])
```

```python
df.to_csv('../data/text_cleaned.csv', index=False)
```
