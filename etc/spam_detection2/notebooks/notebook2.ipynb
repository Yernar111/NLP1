```python
import pandas as pd
df = pd.read_csv("../data/Spam_Email_cleaned.csv", keep_default_na=False)
```

```python
X = df["MESSAGE"]
y = df["CATEGORY"]
print(X.shape)
print(y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(max_features=3000)

X_train = tfidf.fit_transform(X_train) # Обучаем tfidf только на train
X_test = tfidf.transform(X_test)
```

```python
import scipy
scipy.sparse.save_npz('../models/X_features.npz', X_train)
```

```python
from sklearn.naive_bayes import MultinomialNB
```

```python
model1 = MultinomialNB(alpha=1.0) # Чем больше значение alpha, тем сильнее сглаживание.
model1.fit(X_train, y_train)
y_pred1 = model1.predict(X_test)
```

```python
from sklearn.metrics import classification_report, accuracy_score
```

```python
print(f"Accuracy: {accuracy_score(y_test, y_pred1):.4f}")
print(classification_report(y_test, y_pred1))
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model3 = LogisticRegression(max_iter=200)
model3.fit(X_train, y_train)
y_pred3 = model3.predict(X_test)
```

```python
print(f"Accuracy: {accuracy_score(y_test, y_pred3):.4f}")
print(classification_report(y_test, y_pred3))
```
