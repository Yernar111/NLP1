```python
import numpy as np
import tensorflow_hub as hub
```

```python
embed = hub.load("../wiki_words")
```

```python
words = ["king", "man", "woman", "queen"]
vectors = embed(words).numpy()
v_king = vectors[0]
v_man = vectors[1]
v_woman = vectors[2]
v_queen_true = vectors[3]
```

```python
target_vector = v_king - v_man + v_woman
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
sim_to_queen = cosine_similarity([target_vector], [v_queen_true])[0][0]
sim_to_woman = cosine_similarity([target_vector], [v_woman])[0][0]
sim_to_king = cosine_similarity([target_vector], [v_king])[0][0]
sim_to_man = cosine_similarity([target_vector], [v_man])[0][0]
print(sim_to_queen)
print(sim_to_woman)
print(sim_to_king)
print(sim_to_man)
```

```python
word1 = ["kingdom"]
vector1 = embed(word1).numpy()
v_kingdom = vector1[0]
# print(v_kingdom)
print(v_kingdom.shape)
sim_to_kingdom = cosine_similarity([v_king], [v_kingdom])[0][0]
print(sim_to_kingdom)
```
